```python
import pandas as pd
from itertools import combinations
import numpy as np
import matplotlib.pyplot as plt
df_main = pd.read_csv('investor_startup.csv')
```

### Temproal category preference matrix

```python
df_main['announced_on'] = pd.to_datetime(df_main['announced_on'])
df_main['year'] = df_main['announced_on'].dt.year
df_main['year'] = df_main['year'].astype(int)
df_main = df_main.sort_values(by='year', ascending=True)
#df_main['category_region'] = df_main['category_groups_list'] + '-' + df_main['org_region']
df_main = df_main.dropna(subset=['investor_uuid'])

```

```python
df_main['year'].unique()
```

```python
sum = 0
for year in df_main['year'].unique():
    df = df_main[df_main['year'] == year].copy()
    invest = df['investor_uuid'].unique()
    print(year, len(invest))
    sum += len(invest)
    
sum
```

```python
df_yearly = {}

for year in df_main['year'].unique():
    df_year = df_main[df_main['year'] == year].copy()
    df_year['category_list'] = df_year['category_list'].str.split(',')
    expanded_categories = df_year.explode('category_list')
    categories_encoded = expanded_categories['category_list'].str.get_dummies()
    categories_encoded['investor_uuid'] = expanded_categories['investor_uuid']
    
    category_investments = categories_encoded.groupby('investor_uuid').sum()
    category_investments = category_investments.div(category_investments.sum(axis=1), axis=0).fillna(0)
    category_investments['year'] = year
    df_yearly[year] = category_investments

```

```python
df_final = pd.concat(df_yearly.values())
df_final = df_final.fillna(0)
```

```python
df_final
```

```python
df_final.to_csv('investor_category_vector.csv')
```

### Temporal category group preference matrix

```python
df_main['announced_on'] = pd.to_datetime(df_main['announced_on'])
df_main['year'] = df_main['announced_on'].dt.year
df_main['year'] = df_main['year'].astype(int)
df_main = df_main.sort_values(by='year', ascending=True)
#df_main['category_region'] = df_main['category_groups_list'] + '-' + df_main['org_region']
df_main = df_main.dropna(subset=['investor_uuid'])
```

```python
df_yearly = {}

for year in df_main['year'].unique():
    df_year = df_main[df_main['year'] == year].copy()
    df_year['category_groups_list'] = df_year['category_groups_list'].str.split(',')
    expanded_categories = df_year.explode('category_groups_list')
    categories_encoded = expanded_categories['category_groups_list'].str.get_dummies()
    categories_encoded['investor_uuid'] = expanded_categories['investor_uuid']
    
    category_investments = categories_encoded.groupby('investor_uuid').sum()
    category_investments['year'] = year
    df_yearly[year] = category_investments
```

```python
df_final = pd.concat(df_yearly.values())
df_final = df_final.fillna(0)

df_final.to_csv('investor_category_group_temporal_matrix.csv')
```
